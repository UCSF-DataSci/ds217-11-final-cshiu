````python
# Question 1 Wrangled Data
print("="*80)
print("Q3.1: WRANGLED DATA")
print("="*80)

# Load cleaned data
print("\n Loading cleaned data from Q2...")
print("-" * 80)
df = pd.read_csv('output/q2_data_cleaned.csv')
print(f"  Loaded: output/q2_data_cleaned.csv")
print(f"  Shape: {df.shape[0]:,} rows × {df.shape[1]} columns")
print(f"  Columns: {list(df.columns)}")

# Original Datetime column state
print("\n Initial Datetime Column State")
print("-" * 80)
print(f"Datetime column: 'Measurement Timestamp'")
print(f"  Data type: {df['Measurement Timestamp'].dtype}")
print(f"  Sample values:")
print(df['Measurement Timestamp'].head(3))

# Parsing datetime column
print("\n3.1 Parsing Datetime Column")
print("-" * 80)
df['Measurement Timestamp'] = pd.to_datetime(df['Measurement Timestamp'])
print(f"  Converted to datetime using pd.to_datetime()")
print(f"  New data type: {df['Measurement Timestamp'].dtype}")
print(f"  Date range: {df['Measurement Timestamp'].min()} to {df['Measurement Timestamp'].max()}")
time_span = (df['Measurement Timestamp'].max() - df['Measurement Timestamp'].min()).days
print(f"  Time span: {time_span} days")

# Set datetime as index
print("\n3.2 Setting Datetime as Index")
print("-" * 80)
df = df.set_index('Measurement Timestamp')
print(f"  Set 'Measurement Timestamp' as index using df.set_index()")
print(f"  Index name: {df.index.name}")
print(f"  Index type: {type(df.index)}")
print(f"  Is datetime index: {isinstance(df.index, pd.DatetimeIndex)}")

# Sort index chronologically
print("\n3.3 Sorting Index Chronologically")
print("-" * 80)
was_sorted = df.index.is_monotonic_increasing
df = df.sort_index()
print(f"  Sorted using df.sort_index()")
print(f"  Was already sorted: {was_sorted}")
print(f"  Is now sorted: {df.index.is_monotonic_increasing}")
print(f"  First timestamp: {df.index[0]}")
print(f"  Last timestamp: {df.index[-1]}")

# Display wrangled data structure
print("\n Wrangled Data Structure")
print("-" * 80)
print(f"Shape: {df.shape[0]:,} rows x {df.shape[1]} columns")
print(f"Index: DatetimeIndex ('{df.index.name}')")
print(f"Columns ({len(df.columns)}): {list(df.columns)}")
print("\nFirst 5 rows:")
print(df.head())
print("\nLast 5 rows:")
print(df.tail())

# Verify time series properties
print("\n Time Series Verification")
print("-" * 80)
time_diffs = df.index.to_series().diff()
print(f"  Chronologically ordered: {df.index.is_monotonic_increasing}")

# Save with datetime as column (reset index)
print("\n3.4 Saving Wrangled Data with Datetime as Column")
print("-" * 80)
# Reset index to save datetime as a column, not as index
df_to_save = df.reset_index()
print(f"  Reset index to convert datetime back to column")
print(f"  Shape before save: {df_to_save.shape[0]:,} rows × {df_to_save.shape[1]} columns")
print(f"  Columns: {list(df_to_save.columns)}")

# Save to CSV with index=False (no extra index column)
df_to_save.to_csv('output/q3_wrangled_data.csv', index=False)
print(f"\n  Saved to: output/q3_wrangled_data.csv")
print(f"  Format: CSV with datetime as column")
print(f"  Index parameter: index=False (no extra index column)")
````

````python
# Question 2
print("="*80)
print("Q3.2: TEMPORAL FEATURE EXTRACTION")
print("="*80)

# Load wrangled data and set datetime index
print("\n LOADING WRANGLED DATA")
print("-" * 80)
df = pd.read_csv('output/q3_wrangled_data.csv')
print(f"  Loaded: output/q3_wrangled_data.csv")
print(f"  Shape: {df.shape[0]:,} rows × {df.shape[1]} columns")

# Parse datetime and set as index
df['Measurement Timestamp'] = pd.to_datetime(df['Measurement Timestamp'])
df = df.set_index('Measurement Timestamp')
df = df.sort_index()
print(f"  Set datetime index: {df.index.name}")
print(f"  Date range: {df.index.min()} to {df.index.max()}")

# Extract time featuresusing datetime index
print("\n EXTRACTING TEMPORAL FEATURES")
print("-" * 80)

# Required Features
print("Extracting required features:")
df['hour'] = df.index.hour
print(f"  hour: Range {df['hour'].min()}-{df['hour'].max()} (0-23)")

df['day_of_week'] = df.index.dayofweek
print(f"  day_of_week: Range {df['day_of_week'].min()}-{df['day_of_week'].max()} (0=Monday, 6=Sunday)")

df['month'] = df.index.month
print(f"  month: Range {df['month'].min()}-{df['month'].max()} (1-12)")

# Optional but recommended features
print("\nExtracting optional features:")
df['year'] = df.index.year
print(f"  year: Range {df['year'].min()}-{df['year'].max()}")

df['day_name'] = df.index.day_name()
print(f"  day_name: {df['day_name'].unique()}")

df['is_weekend'] = (df.index.dayofweek >= 5).astype(int)
print(f"  is_weekend: {df['is_weekend'].value_counts().to_dict()} (0=weekday, 1=weekend)")

# Display extracted required and optional features
print("\n TEMPORAL FEATURES SUMMARY")
print("-" * 80)
temporal_cols = ['hour', 'day_of_week', 'month', 'year', 'day_name', 'is_weekend']
print(f"Temporal features added: {len(temporal_cols)}")
for col in temporal_cols:
    print(f"\n  {col}:")
    print(f"    Type: {df[col].dtype}")
    print(f"    Missing values: {df[col].isnull().sum()}")
    if col in ['hour', 'day_of_week', 'month', 'year', 'is_weekend']:
        print(f"    Min: {df[col].min()}, Max: {df[col].max()}")
        print(f"    Unique values: {df[col].nunique()}")
    else:  # day_name
        print(f"    Unique values: {df[col].unique()}")

# Show sample data
print("\n SAMPLE DATA WITH TEMPORAL FEATURES")
print("-" * 80)
# Select relevant columns to display
display_cols = temporal_cols
print(f"\nFirst 5 rows (temporal features only):")
print(df[display_cols].head(5))

print(f"\nRandom sample of 5 rows:")
print(df[display_cols].sample(5, random_state=42))

# Validation checks
# Check for NaN in required columns
required_cols = ['hour', 'day_of_week', 'month']
for col in required_cols:
    nan_count = df[col].isnull().sum()
    if nan_count > 0:
        print(f"  {col}: {nan_count} NaN values found")
        validation_passed = False
    else:
        print(f"  {col}: No NaN values")

# Save Temporal Features Data
print("\n Saving Temporal Features Data")
print("-" * 80)

# Reset index to save datetime as column
df_to_save = df.reset_index()

# Verify column order (datetime first, then temporal features, then original data)
cols_order = ['Measurement Timestamp'] + temporal_cols + [col for col in df_to_save.columns 
              if col not in ['Measurement Timestamp'] + temporal_cols]
df_to_save = df_to_save[cols_order]

print(f"Columns to save ({len(df_to_save.columns)}):")
print(f"  {list(df_to_save.columns)}")
print(f"\nFirst column: {df_to_save.columns[0]} (datetime)")
print(f"Temporal features: {temporal_cols}")

# Save to CSV
df_to_save.to_csv('output/q3_temporal_features.csv', index=False)
print(f"\n✓ Saved to: output/q3_temporal_features.csv")
print(f"  Shape: {df_to_save.shape[0]:,} rows × {df_to_save.shape[1]} columns")
print(f"  Index parameter: index=False (no extra index column)")
````

````python
# Question 3 Date Range Information after datetime parsing
print("="*80)
print("Q3.3: GENERATING DATETIME INFO REPORT")
print("="*80)

# Load wrangled data with datetime
print("\n Load Wrangled Data with Datetime")
print("-" * 80)
df = pd.read_csv('output/q3_wrangled_data.csv')
df['Measurement Timestamp'] = pd.to_datetime(df['Measurement Timestamp'])
print(f"  Loaded: output/q3_wrangled_data.csv")
print(f"  Rows: {len(df):,}")

# Get date range information
print("\n Extracting Date Range Information")
print("-" * 80)

start_date = df['Measurement Timestamp'].min()
end_date = df['Measurement Timestamp'].max()
duration = end_date - start_date

print(f"Start date: {start_date}")
print(f"End date: {end_date}")
print(f"Duration: {duration}")

# Calculate detailed duration breakdown
print("\n Calculating Duration Breakdown")
print("-" * 80)

# Calculating year & month
years = end_date.year - start_date.year

print(f"Year range: {years}")


# Calculate duration
duration_parts = []
if years > 0:
    duration_parts.append(f"{years} year{'s' if years != 1 else ''}")

duration_str = ", ".join(duration_parts) if duration_parts else "0 hours"

# Define days/hours conversion
total_days = duration.days
total_hours = duration.total_seconds() / 3600

print(f"\nFormatted duration: {duration_str}")
print(f"Total days: {total_days}")
print(f"Total hours: {total_hours:.1f}")


# Generate report content
report_lines = []
report_lines.append("Date Range After Datetime Parsing:")
report_lines.append(f"Start: {start_date}")
report_lines.append(f"End: {end_date}")
report_lines.append(f"Total Duration: {duration_str}")

report_content = "\n".join(report_lines)

# Save to file
print("-" * 80)
with open('output/q3_datetime_info.txt', 'w') as f:
    f.write(report_content)

print(f" Saved to: output/q3_datetime_info.txt")
````
