### 1. `output/q5_correlations.csv`
**Format:** CSV file
**Content:** Correlation matrix (can be subset of key variables)
**Requirements:**
- Square matrix with variable names as both index and columns
- Values are correlation coefficients (between -1 and 1)
- Can be subset of key variables (e.g., top 10 most important variables)
- **Include index/column names** when saving: `corr_matrix.to_csv('output/q5_correlations.csv')`

**Example format:**
```csv
,Air Temperature,Water Temperature,Wind Speed,Humidity
Air Temperature,1.0,0.847,-0.234,-0.156
Water Temperature,0.847,1.0,0.123,0.089
Wind Speed,-0.234,0.123,1.0,0.456
Humidity,-0.156,0.089,0.456,1.0
```

````python
# Q5.1 Correlation Matrix csv

print("="*80)
print("Q5.1: PATTERN ANALYSIS - CORRELATION MATRIX")
print("="*80)

# Load feature-engineered data from Q4
print("\n LOADING DATA")
print("-" * 80)
df = pd.read_csv('output/q4_features.csv', 
                 parse_dates=['Measurement Timestamp'], 
                 index_col='Measurement Timestamp')
df = df.sort_index()

print(f"  Loaded: output/q4_features.csv")
print(f"  Records: {len(df):,}")
print(f"  Date range: {df.index.min()} to {df.index.max()}")
print(f"  Total columns: {len(df.columns)}")

# Identify numeric columns for correlation analysis
print("\n IDENTIFYING NUMERIC COLUMNS")
print("-" * 80)

numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
print(f"  Found {len(numeric_cols)} numeric columns")

# Identify original sensor/measurement columns (not derived features)
original_cols = [col for col in numeric_cols 
                 if not any(x in col.lower() for x in 
                           ['rolling', 'lag', 'change', 'deviation', 'daily_mean', 
                            'hourly_avg', 'sin', 'cos', '_times_', '_div_', '_minus_'])]

print(f"  Identified {len(original_cols)} original measurement columns")
print(f"\nOriginal columns:")
for col in original_cols:
    print(f"  - {col}")

# Calculate correlation matrix for original columns
print("\n CALCULATING CORRELATION MATRIX (Original Columns)")
print("-" * 80)

if len(original_cols) > 0:
    corr_matrix_original = df[original_cols].corr()
    print(f"  Correlation matrix calculated")
    print(f"  Shape: {corr_matrix_original.shape[0]} x {corr_matrix_original.shape[1]}")
    
    print(f"\nCorrelation matrix preview:")
    print(corr_matrix_original)
    
    # Find strongest correlations (excluding diagonal)
    print(f"\n STRONGEST CORRELATIONS (Original Columns)")
    print("-" * 80)
    
    # Get upper triangle of correlation matrix (avoid duplicates)
    upper_tri = corr_matrix_original.where(
        np.triu(np.ones(corr_matrix_original.shape), k=1).astype(bool)
    )
    
    # Find correlations and sort by absolute value
    correlations = []
    for col in upper_tri.columns:
        for idx in upper_tri.index:
            val = upper_tri.loc[idx, col]
            if pd.notna(val):
                correlations.append({
                    'Variable_1': idx,
                    'Variable_2': col,
                    'Correlation': val,
                    'Abs_Correlation': abs(val)
                })
    
    corr_df = pd.DataFrame(correlations).sort_values('Abs_Correlation', ascending=False)
    
    print(f"Top 10 strongest correlations:")
    for i, row in corr_df.head(10).iterrows():
        print(f"  {row['Variable_1']} <-> {row['Variable_2']}: {row['Correlation']:.3f}")
    
    # Save original correlation matrix
    print(f"\n SAVING CORRELATION MATRIX")
    print("-" * 80)
    
    corr_matrix_original.to_csv('output/q5_correlations.csv')
    print(f"  Saved to: output/q5_correlations.csv")
    print(f"  Shape: {corr_matrix_original.shape[0]} x {corr_matrix_original.shape[1]}")
    print(f"  Includes index and column names: Yes")
    
else:
    print("! No original columns found for correlation analysis")

````

**Format:** PNG image file
**Content:** Advanced visualizations showing trends/seasonality
**Required visualizations (at least 2 of these):**
1. **Trend over time:** Line plot showing variable(s) over time (e.g., monthly averages)

````python
# Q5.2 Advanced Visualizations Showing Trends/Seasonality png

print("="*80)
print("Q5.2: ADVANCED PATTERN VISUALIZATIONS")
print("="*80)

# Set style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (16, 12)

# Load feature-engineered data from Q4
print("\n LOADING DATA")
print("-" * 80)
df = pd.read_csv('output/q4_features.csv', 
                 parse_dates=['Measurement Timestamp'], 
                 index_col='Measurement Timestamp')
df = df.sort_index()

print(f"  Loaded: output/q4_features.csv")
print(f"  Records: {len(df):,}")
print(f"  Date range: {df.index.min()} to {df.index.max()}")

# Identify original numeric columns
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
original_cols = [col for col in numeric_cols 
                 if not any(x in col.lower() for x in 
                           ['rolling', 'lag', 'change', 'deviation', 'daily_mean', 
                            'hourly_avg', 'sin', 'cos', '_times_', '_div_', '_minus_'])]

print(f"  Found {len(original_cols)} original sensor columns")
print(f"  Primary columns: {original_cols[:3]}")

# Select primary variable for detailed analysis
primary_var = original_cols[0] if original_cols else numeric_cols[0]
secondary_var = original_cols[1] if len(original_cols) > 1 else None

print(f"\nAnalyzing patterns for:")
print(f"  - Primary: {primary_var}")
if secondary_var:
    print(f"  - Secondary: {secondary_var}")

# Prepare temporal features for pattern analysis
if 'hour' not in df.columns:
    df['hour'] = df.index.hour
if 'day_of_week' not in df.columns:
    df['day_of_week'] = df.index.dayofweek
if 'month' not in df.columns:
    df['month'] = df.index.month

# Create multi-panel visualization
print("\n CREATING MULTI-PANEL PATTERN VISUALIZATION")
print("-" * 80)

fig = plt.figure(figsize=(16, 12))
gs = fig.add_gridspec(3, 2, hspace=0.3, wspace=0.3)

# PANEL 1: Trend Over Time (Monthly Averages)

print("  Creating Panel 1: Trend Over Time (Monthly)")

ax1 = fig.add_subplot(gs[0, :])

# Resample to monthly averages using month end ("ME")
monthly_data = df[[primary_var]].resample('ME').mean()
if secondary_var:
    monthly_data[secondary_var] = df[secondary_var].resample('ME').mean()

# Plot
ax1.plot(monthly_data.index, monthly_data[primary_var], 
         marker='o', linewidth=2, markersize=6, label=primary_var, color='steelblue')

if secondary_var:
    ax1_twin = ax1.twinx()
    ax1_twin.plot(monthly_data.index, monthly_data[secondary_var], 
                  marker='s', linewidth=2, markersize=6, label=secondary_var, 
                  color='coral', alpha=0.8)
    ax1_twin.set_ylabel(secondary_var, fontsize=11, color='coral')
    ax1_twin.tick_params(axis='y', labelcolor='coral')
    ax1_twin.legend(loc='upper right')

ax1.set_xlabel('Date', fontsize=11)
ax1.set_ylabel(primary_var, fontsize=11, color='steelblue')
ax1.set_title('Trend Over Time: Monthly Averages', fontsize=13, fontweight='bold', pad=15)
ax1.legend(loc='upper left')
ax1.grid(True, alpha=0.3)
ax1.tick_params(axis='y', labelcolor='steelblue')


# PANEL 2: Seasonal Pattern by Month

print("  Creating Panel 2: Seasonal Pattern by Month")

ax2 = fig.add_subplot(gs[1, 0])

monthly_pattern = df.groupby('month')[primary_var].mean()
month_names = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 
               'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

x_pos = np.arange(len(monthly_pattern))
bars = ax2.bar(x_pos, monthly_pattern.values, color='seagreen', alpha=0.7, edgecolor='black')

# Highlight max and min months
max_idx = monthly_pattern.values.argmax()
min_idx = monthly_pattern.values.argmin()
bars[max_idx].set_color('darkgreen')
bars[min_idx].set_color('lightgreen')

ax2.set_xlabel('Month', fontsize=11)
ax2.set_ylabel(f'Average {primary_var}', fontsize=11)
ax2.set_title('Seasonal Pattern: Average by Month', fontsize=13, fontweight='bold', pad=15)
ax2.set_xticks(x_pos)
ax2.set_xticklabels([month_names[i-1] for i in monthly_pattern.index], rotation=45)
ax2.grid(True, alpha=0.3, axis='y')

# Add value labels on bars
for i, (bar, val) in enumerate(zip(bars, monthly_pattern.values)):
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height,
             f'{val:.1f}', ha='center', va='bottom', fontsize=8)


# PANEL 3: Pattern by Day of Week

print("  Creating Panel 3: Pattern by Day of Week")

ax3 = fig.add_subplot(gs[1, 1])

dow_pattern = df.groupby('day_of_week')[primary_var].mean()
day_names = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']

x_pos = np.arange(len(dow_pattern))
bars = ax3.bar(x_pos, dow_pattern.values, color='mediumpurple', alpha=0.7, edgecolor='black')

# Highlight weekends
for i in [5, 6]:  # Saturday and Sunday
    if i in dow_pattern.index:
        idx = list(dow_pattern.index).index(i)
        bars[idx].set_color('purple')

ax3.set_xlabel('Day of Week', fontsize=11)
ax3.set_ylabel(f'Average {primary_var}', fontsize=11)
ax3.set_title('Weekly Pattern: Average by Day of Week', fontsize=13, fontweight='bold', pad=15)
ax3.set_xticks(x_pos)
ax3.set_xticklabels([day_names[i] for i in dow_pattern.index])
ax3.grid(True, alpha=0.3, axis='y')

# Add value labels
for bar, val in zip(bars, dow_pattern.values):
    height = bar.get_height()
    ax3.text(bar.get_x() + bar.get_width()/2., height,
             f'{val:.1f}', ha='center', va='bottom', fontsize=8)


# PANEL 4: Pattern by Hour of Day

print("  Creating Panel 4: Pattern by Hour of Day")

ax4 = fig.add_subplot(gs[2, 0])

hourly_pattern = df.groupby('hour')[primary_var].mean()

ax4.plot(hourly_pattern.index, hourly_pattern.values, 
         marker='o', linewidth=2.5, markersize=5, color='darkorange')
ax4.fill_between(hourly_pattern.index, hourly_pattern.values, alpha=0.3, color='orange')

ax4.set_xlabel('Hour of Day', fontsize=11)
ax4.set_ylabel(f'Average {primary_var}', fontsize=11)
ax4.set_title('Daily Pattern: Average by Hour', fontsize=13, fontweight='bold', pad=15)
ax4.set_xticks(range(0, 24, 2))
ax4.grid(True, alpha=0.3)

# Highlight peak hour
peak_hour = hourly_pattern.idxmax()
peak_value = hourly_pattern.max()
ax4.axvline(x=peak_hour, color='red', linestyle='--', alpha=0.5, label=f'Peak: {peak_hour}:00')
ax4.legend()


# PANEL 5: Correlation Heatmap

print("  Creating Panel 5: Correlation Heatmap")

ax5 = fig.add_subplot(gs[2, 1])

# Select subset of columns for correlation
corr_cols = original_cols[:5] if len(original_cols) >= 5 else original_cols
if len(corr_cols) < 2:
    corr_cols = numeric_cols[:5]

corr_matrix = df[corr_cols].corr()

# Create heatmap
sns.heatmap(corr_matrix, 
            annot=True, 
            fmt='.2f', 
            cmap='coolwarm', 
            center=0, 
            square=True, 
            linewidths=1,
            cbar_kws={'label': 'Correlation'},
            ax=ax5)

ax5.set_title('Correlation Heatmap: Key Variables', fontsize=13, fontweight='bold', pad=15)

# Overall title
fig.suptitle('Time Series Pattern Analysis: Trends and Seasonality', 
             fontsize=16, fontweight='bold', y=0.995)

# Save figure
print("\n3. SAVING VISUALIZATION")
print("-" * 80)

plt.savefig('output/q5_patterns.png', dpi=150, bbox_inches='tight')
print(f"  Saved to: output/q5_patterns.png")
print(f"  Resolution: 150 DPI")
print(f"  Panels: 5 (Trend, Monthly, Weekly, Hourly, Correlation)")
plt.close()


# Summary
print("\n" + "="*80)
print("SUMMARY")
print("="*80)
print(f"  Created multi-panel pattern visualization:")
print(f"  1. Trend over time (monthly averages)")
print(f"  2. Seasonal pattern by month")
print(f"  3. Weekly pattern by day of week")
print(f"  4. Daily pattern by hour")
print(f"  5. Correlation heatmap")
print(f"\n  Output files:")
print(f"  - output/q5_patterns.png (main multi-panel)")

print("\n" + "="*80)
print("PATTERN VISUALIZATION COMPLETE")
print("="*80)
````

```
KEY PATTERNS IDENTIFIED

TEMPORAL TRENDS:

- Higher temperatures in summer months (June-August)
- Lower temperatures in winter months (December-February)
- Monthly air temp range: 4.2°C to 25.8°C

DAILY PATTERNS:

- Peak air temp typically at hour 14-15 (2-3 PM)
- Minimum air temp typically at hour 5-6 (5-6 AM)

CORRELATIONS:
- Air Temp vs Water Temp: 0.847 (strong positive correlation)
- Air Temp vs Humidity: -0.234 (moderate negative correlation)
- Wind Speed vs Wave Height: 0.612 (moderate positive correlation)
```

````python
# Q5.3 Brief Text Summary of Key Patterns Identified - plaint text file

print("="*80)
print("Q5: GENERATING TREND SUMMARY REPORT")
print("="*80)

# Load feature-engineered data
print("\n1. LOADING DATA")
print("-" * 80)
df = pd.read_csv('output/q4_features.csv', 
                 parse_dates=['Measurement Timestamp'], 
                 index_col='Measurement Timestamp')
df = df.sort_index()

print(f"  Loaded: output/q4_features.csv")
print(f"  Records: {len(df):,}")
print(f"  Date range: {df.index.min()} to {df.index.max()}")

# Identify original numeric columns
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
original_cols = [col for col in numeric_cols 
                 if not any(x in col.lower() for x in 
                           ['rolling', 'lag', 'change', 'deviation', 'daily_mean', 
                            'hourly_avg', 'sin', 'cos', '_times_', '_div_', '_minus_'])]

print(f" Analyzing {len(original_cols)} original variables")

# Ensure temporal features exist
if 'hour' not in df.columns:
    df['hour'] = df.index.hour
if 'day_of_week' not in df.columns:
    df['day_of_week'] = df.index.dayofweek
if 'month' not in df.columns:
    df['month'] = df.index.month

# Prepare report content
print("\n ANALYZING PATTERNS")
print("-" * 80)

report_lines = []

def add_line(text="", indent=0):
    """Helper function to add formatted lines"""
    report_lines.append("  " * indent + text)

# Report header
add_line("KEY PATTERNS IDENTIFIED")
add_line("=" * 60)
add_line()


# 1. TEMPORAL TRENDS

print("  Analyzing temporal trends...")

add_line("1. TEMPORAL TRENDS:")
add_line("-" * 60)

# Analyze monthly trends for each variable
monthly_data = df[original_cols].resample('ME').mean()

for col in original_cols[:3]:  # Analyze top 3 variables
    if col in df.columns:
        monthly_vals = df.groupby('month')[col].mean()
        
        # Find seasonal peaks
        max_month = monthly_vals.idxmax()
        min_month = monthly_vals.idxmin()
        max_val = monthly_vals.max()
        min_val = monthly_vals.min()
        range_val = max_val - min_val
        
        month_names = ['January', 'February', 'March', 'April', 'May', 'June',
                       'July', 'August', 'September', 'October', 'November', 'December']
        
        add_line(f"{col}:")
        
        # Determine trend direction
        if range_val > (df[col].std() * 0.5):  # Determines significant seasonal variation
            add_line(f"  - Shows clear seasonal pattern", indent=1)
            add_line(f"  - Peak in {month_names[max_month-1]} (avg: {max_val:.2f})", indent=1)
            add_line(f"  - Lowest in {month_names[min_month-1]} (avg: {min_val:.2f})", indent=1)
            add_line(f"  - Seasonal range: {range_val:.2f}", indent=1)
        else:
            add_line(f"  - Relatively stable across seasons", indent=1)
            add_line(f"  - Mean: {df[col].mean():.2f}, Std: {df[col].std():.2f}", indent=1)
        
        # Overall trend (first half vs second half)
        midpoint = df.index[len(df)//2]
        first_half_mean = df.loc[:midpoint, col].mean()
        second_half_mean = df.loc[midpoint:, col].mean()
        trend_diff = second_half_mean - first_half_mean
        
        if abs(trend_diff) > df[col].std() * 0.2:
            direction = "increasing" if trend_diff > 0 else "decreasing"
            add_line(f"  - Overall trend: {direction} ({trend_diff:+.2f})", indent=1)
        else:
            add_line(f"  - Overall trend: stable", indent=1)
        
        add_line()


# DAILY PATTERNS

print("  Analyzing daily patterns...")

add_line("2a. DAILY PATTERNS:")
add_line("-" * 60)

for col in original_cols[:2]:  # Analyze top 2 variables
    if col in df.columns:
        hourly_pattern = df.groupby('hour')[col].mean()
        
        peak_hour = hourly_pattern.idxmax()
        min_hour = hourly_pattern.idxmin()
        peak_val = hourly_pattern.max()
        min_val = hourly_pattern.min()
        daily_range = peak_val - min_val
        
        add_line(f"{col}:")
        
        if daily_range > df[col].std() * 0.3:  # Significant daily variation
            add_line(f"  - Shows diurnal cycle (day/night variation)", indent=1)
            add_line(f"  - Peak at hour {peak_hour}:00 ({peak_hour % 12 if peak_hour != 12 else 12} {'PM' if peak_hour >= 12 else 'AM'}): {peak_val:.2f}", indent=1)
            add_line(f"  - Minimum at hour {min_hour}:00 ({min_hour % 12 if min_hour != 12 else 12} {'PM' if min_hour >= 12 else 'AM'}): {min_val:.2f}", indent=1)
            add_line(f"  - Daily range: {daily_range:.2f}", indent=1)
        else:
            add_line(f"  - Relatively stable throughout the day", indent=1)
            add_line(f"  - Average: {hourly_pattern.mean():.2f}", indent=1)
        
        add_line()


# WEEKLY PATTERNS

print("  Analyzing weekly patterns...")

add_line("2b. WEEKLY PATTERNS:")
add_line("-" * 60)

for col in original_cols[:2]:
    if col in df.columns:
        dow_pattern = df.groupby('day_of_week')[col].mean()
        
        weekday_mean = dow_pattern.iloc[:5].mean()  # Mon-Fri
        weekend_mean = dow_pattern.iloc[5:].mean()  # Sat-Sun
        
        diff = abs(weekend_mean - weekday_mean)
        
        add_line(f"{col}:")
        if diff > df[col].std() * 0.1:
            higher = "weekend" if weekend_mean > weekday_mean else "weekday"
            add_line(f"  - Shows weekly pattern with higher values on {higher}s", indent=1)
            add_line(f"  - Weekday average: {weekday_mean:.2f}", indent=1)
            add_line(f"  - Weekend average: {weekend_mean:.2f}", indent=1)
            add_line(f"  - Difference: {diff:.2f}", indent=1)
        else:
            add_line(f"  - No significant weekday vs weekend difference", indent=1)
            add_line(f"  - Consistent throughout the week", indent=1)
        
        add_line()

# MONTHLY PATTERNS

print("  Analyzing monthly patterns...")

add_line("2c. MONTHLY PATTERNS:")
add_line("-" * 60)

for col in original_cols[:2]:
    if col in df.columns:
        monthly_pattern = df.groupby('month')[col].mean()
        
        # Define seasons
        winter_months = [12, 1, 2]  # Dec, Jan, Feb
        spring_months = [3, 4, 5]   # Mar, Apr, May
        summer_months = [6, 7, 8]   # Jun, Jul, Aug
        fall_months = [9, 10, 11]   # Sep, Oct, Nov
        
        # Calculate seasonal averages
        winter_mean = monthly_pattern[monthly_pattern.index.isin(winter_months)].mean()
        spring_mean = monthly_pattern[monthly_pattern.index.isin(spring_months)].mean()
        summer_mean = monthly_pattern[monthly_pattern.index.isin(summer_months)].mean()
        fall_mean = monthly_pattern[monthly_pattern.index.isin(fall_months)].mean()
        
        # Find peak and low months
        peak_month = monthly_pattern.idxmax()
        low_month = monthly_pattern.idxmin()
        month_names = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 
                       'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
        
        seasonal_range = monthly_pattern.max() - monthly_pattern.min()
        
        add_line(f"{col}:")
        if seasonal_range > df[col].std() * 0.3:  # Significant seasonal variation
            add_line(f"  - Shows strong seasonal pattern", indent=1)
            add_line(f"  - Peak month: {month_names[peak_month-1]} (avg: {monthly_pattern.max():.2f})", indent=1)
            add_line(f"  - Lowest month: {month_names[low_month-1]} (avg: {monthly_pattern.min():.2f})", indent=1)
            add_line(f"  - Seasonal range: {seasonal_range:.2f}", indent=1)
            add_line()
            add_line(f"  Seasonal breakdown:", indent=1)
            add_line(f"    • Winter (Dec-Feb): {winter_mean:.2f}", indent=1)
            add_line(f"    • Spring (Mar-May): {spring_mean:.2f}", indent=1)
            add_line(f"    • Summer (Jun-Aug): {summer_mean:.2f}", indent=1)
            add_line(f"    • Fall (Sep-Nov): {fall_mean:.2f}", indent=1)
        else:
            add_line(f"  - No significant seasonal variation", indent=1)
            add_line(f"  - Relatively stable across months (avg: {monthly_pattern.mean():.2f})", indent=1)
            add_line(f"  - Monthly range: {seasonal_range:.2f}", indent=1)
        
        add_line()


# CORRELATIONS

print("  Analyzing correlations...")

add_line("3. CORRELATIONS:")
add_line("-" * 60)

if len(original_cols) > 1:
    corr_matrix = df[original_cols].corr()
    
    # Get upper triangle (avoid duplicates and self-correlations)  !!!!!!!!!!!!
    upper_tri = corr_matrix.where(
        np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
    )
    
    # Find top correlations
    correlations = []
    for col in upper_tri.columns:
        for idx in upper_tri.index:
            val = upper_tri.loc[idx, col]
            if pd.notna(val):
                correlations.append({
                    'var1': idx,
                    'var2': col,
                    'corr': val,
                    'abs_corr': abs(val)
                })
    
    corr_df = pd.DataFrame(correlations).sort_values('abs_corr', ascending=False)
    
    # Report top 3-5 correlations
    top_n = min(5, len(corr_df))
    
    for i, row in corr_df.head(top_n).iterrows():
        corr_val = row['corr']
        
        # Determine strength
        abs_corr = abs(corr_val)
        if abs_corr > 0.7:
            strength = "strong"
        elif abs_corr > 0.4:
            strength = "moderate"
        else:
            strength = "weak"
        
        direction = "positive" if corr_val > 0 else "negative"
        
        add_line(f"- {row['var1']} vs {row['var2']}: {corr_val:.3f}")
        add_line(f"  ({strength} {direction} correlation)", indent=1)
    
    add_line()
else:
    add_line("- Insufficient variables for correlation analysis")
    add_line()


# SUMMARY STATISTICS

print("  Calculating summary statistics...")

add_line("SUMMARY STATISTICS:")
add_line("-" * 60)

for col in original_cols[:3]:
    if col in df.columns:
        add_line(f"{col}:")
        add_line(f"  - Mean: {df[col].mean():.2f}", indent=1)
        add_line(f"  - Std Dev: {df[col].std():.2f}", indent=1)
        add_line(f"  - Min: {df[col].min():.2f}", indent=1)
        add_line(f"  - Max: {df[col].max():.2f}", indent=1)
        add_line(f"  - Range: {df[col].max() - df[col].min():.2f}", indent=1)
        add_line()

# Report footer
add_line("=" * 60)
add_line(f"Analysis Period: {df.index.min().date()} to {df.index.max().date()}")
add_line(f"Total Records: {len(df):,}")
add_line(f"Variables Analyzed: {len(original_cols)}")
add_line("=" * 60)

# Write report to file
print("\n3. SAVING REPORT")
print("-" * 80)

report_content = "\n".join(report_lines)

with open('output/q5_trend_summary.txt', 'w') as f:
    f.write(report_content)

print(f"  Saved to: output/q5_trend_summary.txt")
print(f"  Report length: {len(report_lines)} lines")

# Display report
print("\n" + "="*80)
print("REPORT PREVIEW")
print("="*80)
print(report_content)


# Summary
print("\n" + "="*80)
print("SUMMARY")
print("="*80)
print(f"  Generated comprehensive trend summary report")
print(f"  Analyzed:")
print(f"  - Temporal trends (seasonal, overall direction)")
print(f"  - Daily patterns (hourly cycles)")
print(f"  - Weekly patterns (weekday vs weekend)")
print(f"  - Correlations (top 5 relationships)")
print(f"  - Summary statistics")
print(f"\n✓ Saved to: output/q5_trend_summary.txt")

print("\n" + "="*80)
print("TREND SUMMARY GENERATION COMPLETE")
print("="*80)
````
